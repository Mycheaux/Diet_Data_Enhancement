# Supervised Prediction

This notebook loads a preprocessed `X` table, aligns it with a TRE outcome table `Y`, and evaluates simple scikit-learn models using five repeated 80/20 train-test splits.

Classification metrics: AUROC, AUPRC, F1, accuracy.

Regression metrics: R2, RMSE, Pearson correlation.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path('..').resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from downstream_analysis.modeling import run_prediction_from_files, read_table

PROJECT_ROOT
```

## User Settings

Choose `X_PATH`, `Y_PATH`, the target column, task type, and models. Set `TASK_TYPE = None` to infer binary/multiclass/continuous from the target.

```python
X_PATH = PROJECT_ROOT / 'downstream_analysis' / 'outputs' / 'X_embedding_participant.parquet'
Y_PATH = PROJECT_ROOT / 'tre_inputs' / 'outcomes.csv'

TARGET_COL = 'target_variable_name'
ID_COL = 'participant_id'
TIME_COL = 'time_window'  # set to None to align only by participant_id

# Choose: None, 'binary', 'multiclass', or 'continuous'
TASK_TYPE = None

# For classification: logistic_regression, rf_classification
# For regression: linear_regression, ridge, lasso, rf_regression
MODELS = None  # None runs all models available for inferred task type

N_SPLITS = 5
TEST_SIZE = 0.20

OUTPUT_DIR = PROJECT_ROOT / 'downstream_analysis' / 'outputs' / 'prediction_results' / TARGET_COL
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

## Run Prediction

```python
summary = run_prediction_from_files(
    x_path=X_PATH,
    y_path=Y_PATH,
    target_col=TARGET_COL,
    output_dir=OUTPUT_DIR,
    id_col=ID_COL,
    time_col=TIME_COL,
    task_type=TASK_TYPE,
    models=MODELS,
    n_splits=N_SPLITS,
    test_size=TEST_SIZE,
)
summary
```

## Metrics

```python
summary_metrics = read_table(summary['summary_metrics'])
fold_metrics = read_table(summary['fold_metrics'])
display(summary_metrics)
display(fold_metrics.head())
```
